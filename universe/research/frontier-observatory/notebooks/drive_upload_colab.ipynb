# Upload Paper 1 v2.0 to Google Drive (Easy)

**Easiest method:** Run this in Colab, authenticates automatically


```python
# Authenticate and mount Google Drive
from google.colab import auth
from google.colab import drive

auth.authenticate_user()
drive.mount('/content/drive')
print("✓ Google Drive authenticated")
```

```python
# Clone repo and get PDF
import subprocess
import os
from shutil import copy

os.chdir('/tmp')
print("Cloning Tofoo- repo...")
result = subprocess.run(
    ['git', 'clone', '--quiet', '--depth', '1', 
     'https://github.com/nsolland/Tofoo-.git'],
    capture_output=True
)

if result.returncode != 0:
    print("Note: Repo is private, trying alternative...")
    print("If cloning fails, upload PDF manually from your local disk")
else:
    print("✓ Repo cloned")

pdf_file = '/tmp/Tofoo-/Paper1_v2.0_External_Public.pdf'
if os.path.exists(pdf_file):
    print(f"✓ PDF found: {os.path.getsize(pdf_file)/1024:.0f} KB")
else:
    print(f"✗ PDF not found - you can upload manually to Google Drive")
```

```python
# Upload to Drive
import os
from shutil import copy

pdf_file = '/tmp/Tofoo-/Paper1_v2.0_External_Public.pdf'
drive_folder = '/content/drive/MyDrive/Tofoo'

# Create folder
os.makedirs(drive_folder, exist_ok=True)

if os.path.exists(pdf_file):
    dest = os.path.join(drive_folder, 'Paper1_v2.0_External_Public.pdf')
    copy(pdf_file, dest)
    print(f"✓ UPLOAD SUCCESSFUL!")
    print(f"  Location: /MyDrive/Tofoo/Paper1_v2.0_External_Public.pdf")
else:
    print("PDF not available - manual upload needed")
    print("Open Files app in Colab → upload Paper1_v2.0_External_Public.pdf")
```

```python
# Verify in Drive
import os

drive_file = '/content/drive/MyDrive/Tofoo/Paper1_v2.0_External_Public.pdf'
if os.path.exists(drive_file):
    size = os.path.getsize(drive_file) / 1024
    print(f"✓ File verified in Google Drive")
    print(f"  Size: {size:.0f} KB")
    print(f"  Path: /MyDrive/Tofoo/Paper1_v2.0_External_Public.pdf")
else:
    print("File not found in Drive")
```
