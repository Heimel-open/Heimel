# Upload Paper 1 v2.0 External to Google Drive

**One-click upload:** Run cells below to convert markdown to PDF and save to Drive.


```python
# Authenticate with Google Drive
from google.colab import auth
from google.colab import drive

auth.authenticate_user()
drive.mount('/content/drive')
print("✓ Google Drive mounted")
```

```python
# Clone repo (private — you may need to authenticate)
import subprocess
import os

os.chdir('/tmp')
result = subprocess.run(
    ['git', 'clone', 'https://github.com/nsolland/Tofoo-.git'],
    capture_output=True, text=True
)

if result.returncode == 0:
    print("✓ Repository cloned")
else:
    print(f"Could not clone (private repo). Enter path to local file:")
    print(f"Error: {result.stderr}")

# Check if file exists
markdown_file = '/tmp/Tofoo-/theory/2026-07-02-paper1-framleis-law-external-v2-public.md'
if os.path.exists(markdown_file):
    print(f"✓ Found: {markdown_file}")
else:
    print(f"✗ File not found at {markdown_file}")
```

```python
# Install pandoc and other dependencies
import subprocess

subprocess.run(['apt-get', 'update'], capture_output=True)
subprocess.run(['apt-get', 'install', '-y', 'pandoc', 'texlive', 'texlive-latex-extra'], 
               capture_output=True)
print("✓ Pandoc and LaTeX installed")
```

```python
# Convert markdown to PDF
import subprocess
import os

markdown_file = '/tmp/Tofoo-/theory/2026-07-02-paper1-framleis-law-external-v2-public.md'
pdf_file = '/tmp/Paper1_v2.0_External_Public.pdf'

if os.path.exists(markdown_file):
    # Convert using pandoc
    result = subprocess.run([
        'pandoc',
        markdown_file,
        '-o', pdf_file,
        '--from=markdown',
        '--to=pdf',
        '--toc',
        '--number-sections',
        '-V', 'mainfont:DejaVuSans',
        '-V', 'fontsize:11pt',
        '-V', 'margin-top:2cm',
        '-V', 'margin-bottom:2cm',
    ], capture_output=True, text=True)
    
    if result.returncode == 0 and os.path.exists(pdf_file):
        size_mb = os.path.getsize(pdf_file) / (1024*1024)
        print(f"✓ PDF created: {pdf_file}")
        print(f"  Size: {size_mb:.2f} MB")
    else:
        print(f"✗ Conversion failed: {result.stderr}")
else:
    print(f"✗ Markdown file not found")
```

```python
# Upload to Google Drive
from google.colab import drive
from shutil import copy
import os

pdf_file = '/tmp/Paper1_v2.0_External_Public.pdf'
drive_path = '/content/drive/MyDrive/Tofoo'

# Create folder if it doesn't exist
os.makedirs(drive_path, exist_ok=True)

if os.path.exists(pdf_file):
    dest = os.path.join(drive_path, 'Paper1_v2.0_External_Public.pdf')
    copy(pdf_file, dest)
    print(f"✓ Uploaded to: {dest}")
    print(f"\nDrive path: /MyDrive/Tofoo/Paper1_v2.0_External_Public.pdf")
else:
    print(f"✗ PDF file not found")
```

```python
# Verify upload
import os

drive_path = '/content/drive/MyDrive/Tofoo'
pdf_name = 'Paper1_v2.0_External_Public.pdf'

if os.path.exists(os.path.join(drive_path, pdf_name)):
    size = os.path.getsize(os.path.join(drive_path, pdf_name))
    print(f"✓ UPLOAD SUCCESSFUL")
    print(f"  File: {pdf_name}")
    print(f"  Size: {size/(1024*1024):.2f} MB")
    print(f"  Location: /MyDrive/Tofoo/{pdf_name}")
else:
    print(f"✗ Upload verification failed")
```
